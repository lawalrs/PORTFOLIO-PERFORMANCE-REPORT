```python
#Task 1 Variables
Investor_Name = "Chinedu Okeke"
Investment_Name = "Tech Growth Fund"
Initial_Investment = 2500000
Current_Portfolio_Value = 2875000
Additional_Deposit_Today = 150000
Target_Portfolio_Value = 3000000
Annual_Target_Return = 15
Current_Return = 13
Portfolio_Active = True
print(type(Investor_Name))
print(type(Investment_Name))
print(type(Initial_Investment))    
print(type(Current_Portfolio_Value))
print(type(Target_Portfolio_Value))
print(type(Annual_Target_Return))
print(type(Current_Return))
print(type(Portfolio_Active)) 
```

```python
#Task 2 Current Investment Profit
Current_Profit = Current_Portfolio_Value - Initial_Investment
print(Current_Profit)
```

```python
#Task 3 Return on Investment
ROI = (Current_Profit/Initial_Investment)*100
print(ROI)
```

```python
#Task 4 Updated Portfolio Value
Additoinal_Deposit = 150000
Updated_Portfolio_Value = Current_Portfolio_Value + Additoinal_Deposit
Current_Portfolio_Value = 2875000
print(Updated_Portfolio_Value)
```

```python
#Task 5 Investment Target
Investment_Target = Updated_Portfolio_Value >= Target_Portfolio_Value
print(Investment_Target)
```

```python
#Task 6 Investment Performance
Investment_Performance = Current_Return >= Annual_Target_Return
print(Investment_Performance)
```

```python
#Task 7 Portfo;io Health
Portfolio_Health = Portfolio_Active and Target_Portfolio_Value >= 3000000
print(Portfolio_Health)
```

```python
#Task 8 Financial Advisor Review
print(not Portfolio_Active or not Target_Portfolio_Value >= 3000000)
```
