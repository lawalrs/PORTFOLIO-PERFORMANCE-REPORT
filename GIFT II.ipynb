```python
# Task 11: Create Loan Eligibility Function

def check_loan_eligibility(
    age,
    monthly_income,
    debt_to_income_ratio,
    credit_score,
    existing_loan,
    employment_status,
    loan_amount,
    loan_term
):
   
    if age < 18 or age > 65:
        return "Not Eligible: Invalid age"  
    if monthly_income < 150000:
        return "Not Eligible: Income is below ₦150,000"
    if debt_to_income_ratio > 40:
        return "Not Eligible: Debt-to-Income Ratio is above 40%"
    if credit_score < 650:
        return "Not Eligible: Credit score is below 650" 
    if existing_loan == True:
        return "Not Eligible: Customer has an existing loan"
    if employment_status == "Unemployed":
        return "Not Eligible: Customer is unemployed"
    elif employment_status == "Employed" or employment_status == "Self-Employed":
        pass
    else:
        return "Not Eligible: Invalid employment status"
    if loan_amount <= 0 or loan_amount > 5000000:
        return "Not Eligible: Invalid loan amount"
    if loan_term != 6 and loan_term != 12 and loan_term != 24:
        return "Not Eligible: Invalid loan term"
    return "Eligible for Loan"
debt_to_income_ratio = (existing_monthly_debt / monthly_income) * 100

eligibility_result = check_loan_eligibility(
    age,
    monthly_income,
    debt_to_income_ratio,
    credit_score,
    existing_loan,
    employment_status,
    requested_loan_amount,
    loan_term
)

# Display result
print("Customer:", customer_name)
print("Loan Eligibility:", eligibility_result)
```

```python
# ==========================================
# BONUS CHALLENGE
# PROCESS FOUR LOAN APPLICATIONS
# ==========================================

loan_applications = [
    {
        "name": "Daniel Okafor",
        "age": 35,
        "monthly_income": 350000,
        "existing_monthly_debt": 70000,
        "loan_amount": 1000000,
        "loan_term": 12,
        "credit_score": 720,
        "existing_loan": False,
        "employment_status": "Employed"
    },
    {
        "name": "Grace Bello",
        "age": 42,
        "monthly_income": 500000,
        "existing_monthly_debt": 100000,
        "loan_amount": 2000000,
        "loan_term": 24,
        "credit_score": 780,
        "existing_loan": False,
        "employment_status": "Employed"
    },
    {
        "name": "Samuel Ade",
        "age": 27,
        "monthly_income": 120000,
        "existing_monthly_debt": 20000,
        "loan_amount": 500000,
        "loan_term": 12,
        "credit_score": 700,
        "existing_loan": False,
        "employment_status": "Employed"
    },
    {
        "name": "Anita James",
        "age": 50,
        "monthly_income": 400000,
        "existing_monthly_debt": 200000,
        "loan_amount": 1500000,
        "loan_term": 12,
        "credit_score": 680,
        "existing_loan": True,
        "employment_status": "Employed"
    }
]


# ==========================================
# VARIABLES FOR SUMMARY
# ==========================================

eligible_count = 0
not_eligible_count = 0

highest_credit_score = 0
highest_credit_applicant = ""

highest_income = 0
highest_income_applicant = ""

highest_dti = 0
highest_dti_applicant = ""

lowest_dti = float("inf")
lowest_dti_applicant = ""

highest_loan_amount = 0
highest_loan_applicant = ""

income_failed_applicant = ""
existing_loan_applicant = ""


# ==========================================
# PROCESS APPLICATIONS USING FOR LOOP
# ==========================================

for applicant in loan_applications:

    name = applicant["name"]
    age = applicant["age"]
    income = applicant["monthly_income"]
    debt = applicant["existing_monthly_debt"]
    loan_amount = applicant["loan_amount"]
    loan_term = applicant["loan_term"]
    credit_score = applicant["credit_score"]
    existing_loan = applicant["existing_loan"]
    employment_status = applicant["employment_status"]

    # Calculate Debt-to-Income Ratio
    dti = (debt / income) * 100

    # Eligibility checks
    valid_age = age >= 18 and age <= 65
    valid_income = income >= 150000
    valid_loan_amount = loan_amount > 0 and loan_amount <= 5000000
    valid_credit_score = credit_score >= 650
    valid_dti = dti <= 40
    valid_existing_loan = existing_loan == False
    valid_employment = (
        employment_status == "Employed"
        or employment_status == "Self-Employed"
    )
    valid_loan_term = (
        loan_term == 6
        or loan_term == 12
        or loan_term == 24
    )

    # Overall eligibility
    eligible = (
        valid_age
        and valid_income
        and valid_loan_amount
        and valid_credit_score
        and valid_dti
        and valid_existing_loan
        and valid_employment
        and valid_loan_term
    )

    # Count eligible and not eligible applicants
    if eligible:
        eligible_count += 1
    else:
        not_eligible_count += 1

    # Find highest credit score
    if credit_score > highest_credit_score:
        highest_credit_score = credit_score
        highest_credit_applicant = name

    # Find highest income
    if income > highest_income:
        highest_income = income
        highest_income_applicant = name

    # Find highest DTI
    if dti > highest_dti:
        highest_dti = dti
        highest_dti_applicant = name

    # Find lowest DTI
    if dti < lowest_dti:
        lowest_dti = dti
        lowest_dti_applicant = name

    # Find highest loan amount
    if loan_amount > highest_loan_amount:
        highest_loan_amount = loan_amount
        highest_loan_applicant = name

    # Find applicant who failed income requirement
    if not valid_income:
        income_failed_applicant = name

    # Find applicant with existing loan
    if existing_loan:
        existing_loan_applicant = name


# ==========================================
# DISPLAY SUMMARY
# ==========================================

print("=" * 50)
print("       LOAN APPLICATION SUMMARY")
print("=" * 50)

print("Eligible Applicants:", eligible_count)
print("Not Eligible Applicants:", not_eligible_count)

print(
    "Highest Credit Score:",
    highest_credit_applicant,
    "-", highest_credit_score
)

print(
    "Highest Monthly Income:",
    highest_income_applicant,
    f"- ₦{highest_income:,.0f}"
)

print(
    "Highest Debt-to-Income Ratio:",
    highest_dti_applicant,
    f"- {highest_dti:.2f}%"
)

print(
    "Lowest Debt-to-Income Ratio:",
    lowest_dti_applicant,
    f"- {lowest_dti:.2f}%"
)

print(
    "Failed Income Requirement:",
    income_failed_applicant
)

print(
    "Failed Because of Existing Loan:",
    existing_loan_applicant
)

print(
    "Highest Requested Loan:",
    highest_loan_applicant,
    f"- ₦{highest_loan_amount:,.0f}"
)


# ==========================================
# MONTHLY PAYMENT FOR ELIGIBLE APPLICANTS
# ==========================================

print("\nEstimated Monthly Payment for Eligible Applicants:")

for applicant in loan_applications:

    # Calculate DTI again
    dti = (
        applicant["existing_monthly_debt"]
        / applicant["monthly_income"]
    ) * 100

    # Check eligibility
    eligible = (
        applicant["age"] >= 18
        and applicant["age"] <= 65
        and applicant["monthly_income"] >= 150000
        and applicant["loan_amount"] > 0
        and applicant["loan_amount"] <= 5000000
        and applicant["credit_score"] >= 650
        and dti <= 40
        and applicant["existing_loan"] == False
        and (
            applicant["employment_status"] == "Employed"
            or applicant["employment_status"] == "Self-Employed"
        )
        and (
            applicant["loan_term"] == 6
            or applicant["loan_term"] == 12
            or applicant["loan_term"] == 24
        )
    )

    if eligible:
        monthly_payment = (
            applicant["loan_amount"]
            / applicant["loan_term"]
        )

        print(
            applicant["name"],
            f"- ₦{monthly_payment:,.2f}"
        )
```
